```python
# Tokenization Examples in Python
# Author: Phd Abdelouaheb (summary from Utkarsh Kant article)

# -------------------------------
# 1. Using split()
# -------------------------------
text = "Natural Language Processing is amazing! Let's tokenize this text."
print("Split on whitespace:", text.split())

# Sentence splitting
sentences = text.split(".")
print("Split into sentences:", [s.strip() for s in sentences if s])

# -------------------------------
# 2. Using NLTK
# -------------------------------
import nltk

# Download the required tokenizers
nltk.download("punkt")       # older versions
nltk.download("punkt_tab")   # newer versions may need this

from nltk.tokenize import word_tokenize, sent_tokenize, WordPunctTokenizer

# Word tokenization
print("NLTK word_tokenize:", word_tokenize(text))

# Sentence tokenization
print("NLTK sent_tokenize:", sent_tokenize(text))

# WordPunctTokenizer
wp_tokenizer = WordPunctTokenizer()
print("NLTK WordPunctTokenizer:", wp_tokenizer.tokenize(text))

# -------------------------------
# 3. Using spaCy
# -------------------------------


import spacy

# Download the model if missing
import spacy.cli
spacy.cli.download("en_core_web_sm")

# Load it
nlp = spacy.load("en_core_web_sm")
doc = nlp(text)

# Word tokens
print("spaCy word tokens:", [token.text for token in doc])

# Sentence tokens
print("spaCy sentence tokens:", [sent.text for sent in doc.sents])

# Entities (bonus)
print("Named Entities:", [(ent.text, ent.label_) for ent in doc.ents])
```
