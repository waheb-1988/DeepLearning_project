```python
# ---------------------------------------------------------
# Stop Words Removal using Different Libraries in Python
# ---------------------------------------------------------

# Sample text
text = "When I first met her she was very quiet. She remained quiet during the entire two hour long journey from Stony Brook to New York."

# -------------------------------
# 1. NLTK
# -------------------------------
import nltk
from nltk.corpus import stopwords

# Make sure stopwords are downloaded
nltk.download("stopwords")

sw_nltk = stopwords.words("english")
print("NLTK stop words count:", len(sw_nltk))  # ~179

words_nltk = [word for word in text.split() if word.lower() not in sw_nltk]
print("NLTK Result:", " ".join(words_nltk))
print("-" * 50)

# -------------------------------
# 2. spaCy
# -------------------------------
import spacy
import spacy.cli

# Ensure model is available
try:
    nlp = spacy.load("en_core_web_sm")
except OSError:
    spacy.cli.download("en_core_web_sm")
    nlp = spacy.load("en_core_web_sm")

sw_spacy = nlp.Defaults.stop_words
print("spaCy stop words count:", len(sw_spacy))  # ~326

words_spacy = [word for word in text.split() if word.lower() not in sw_spacy]
print("spaCy Result:", " ".join(words_spacy))
print("-" * 50)

# -------------------------------
# 3. Gensim
# -------------------------------
from gensim.parsing.preprocessing import remove_stopwords, STOPWORDS

print("Gensim stop words count:", len(STOPWORDS))  # ~337
print("Gensim Result:", remove_stopwords(text))
print("-" * 50)

# -------------------------------
# 4. Scikit-Learn
# -------------------------------
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

print("Scikit-Learn stop words count:", len(ENGLISH_STOP_WORDS))  # ~318

words_sklearn = [word for word in text.split() if word.lower() not in ENGLISH_STOP_WORDS]
print("Scikit-Learn Result:", " ".join(words_sklearn))
print("-" * 50)

# -------------------------------
# 5. Custom Stop Words
# -------------------------------
my_stop_words = ["her", "me", "i", "she", "it"]

words_custom = [word for word in text.split() if word.lower() not in my_stop_words]
print("Custom Stop Words Result:", " ".join(words_custom))

# Example: Add to NLTK list
sw_nltk.extend(["first", "second"])
print("Extended NLTK stop words count:", len(sw_nltk))

# Example: Remove from NLTK list
if "not" in sw_nltk:
    sw_nltk.remove("not")
print("'not' removed from NLTK stop words")
```
