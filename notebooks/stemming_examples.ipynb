```python
# ---------------------------------------------------------
# NLP Stemming Examples using NLTK
# ---------------------------------------------------------

import nltk
from nltk.stem import PorterStemmer, SnowballStemmer, LancasterStemmer

# Make sure resources are available
nltk.download("punkt")

# Sample text
words = ["flying", "flies", "waited", "waiting", "universal", "university", "universe", "alumnus", "alumni", "alumnae", "caring", "was"]

# -------------------------------
# 1. Porter Stemmer
# -------------------------------
porter = PorterStemmer()
porter_results = [porter.stem(word) for word in words]
print("Porter Stemmer Results:")
print(dict(zip(words, porter_results)))
print("-" * 50)

# -------------------------------
# 2. Snowball Stemmer
# -------------------------------
snowball = SnowballStemmer("english")
snowball_results = [snowball.stem(word) for word in words]
print("Snowball Stemmer Results:")
print(dict(zip(words, snowball_results)))
print("-" * 50)

# -------------------------------
# 3. Lancaster Stemmer
# -------------------------------
lancaster = LancasterStemmer()
lancaster_results = [lancaster.stem(word) for word in words]
print("Lancaster Stemmer Results:")
print(dict(zip(words, lancaster_results)))
print("-" * 50)

# -------------------------------
# Comparison Summary
# -------------------------------
import pandas as pd

df = pd.DataFrame({
    "Word": words,
    "Porter": porter_results,
    "Snowball": snowball_results,
    "Lancaster": lancaster_results
})

print("Comparison Table:")
print(df)
```
