```python

# Regular Expressions in Python
# Author: Phd Abdelouaheb

import re
import pandas as pd

# === Basic Regex Examples ===
text = "My phone number is +123-4567 and my email is test123@example.com"

print("\n=== Digits ===")
print(re.findall(r"\d+", text))   # ['123', '4567', '123']

print("\n=== Words ===")
print(re.findall(r"\w+", text))   # ['My', 'phone', 'number', ...]

print("\n=== Emails ===")
print(re.findall(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}", text))

# === Example 1: Splitting Names ===
data1 = {"Full_Name": ["Mr John Doe", "Dr Jane Smith", "Miss Sarah Connor"]}
df1 = pd.DataFrame(data1)
pattern1 = r"(Mr|Ms|Miss|Mrs|Dr|Alhaji|Sis)\s+(\w+)\s+(\w+)"
df1[["Title", "First_Name", "Surname"]] = df1["Full_Name"].str.extract(pattern1)
print("\n=== Names ===")
print(df1)

# === Example 2: Phone Numbers ===
data2 = {"Phone": ["+123-4567", "+234-9876", "+44-1234"]}
df2 = pd.DataFrame(data2)
pattern2 = r"(\+\d+)\-(\d+)"
df2[["Area_Code", "Line_Number"]] = df2["Phone"].str.extract(pattern2)
print("\n=== Phone Numbers ===")
print(df2)

# === Example 3: Addresses ===
data3 = {"Address": ["12 King, London", "34 Queen, Paris", "56 Main, Berlin"]}
df3 = pd.DataFrame(data3)
pattern3 = r"(\w+)\s+(\w+)\,\s+(\w+)"
df3[["House_No", "Street", "City"]] = df3["Address"].str.extract(pattern3)
print("\n=== Addresses ===")
print(df3)
```
