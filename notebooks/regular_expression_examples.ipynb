```python
# ==================================================
# Regular Expressions in Python
# Author: Phd Abdelouaheb
# ==================================================
# https://regex101.com/
import re
import pandas as pd

# --------------------------------------------------
# === Basic Regex Examples ===
# --------------------------------------------------
text = "My phone number is +123-4567 and my email is test123@example.com"

print("\n=== Digits ===")
print(re.findall(r"\d+", text))   # ['123', '4567', '123']

print("\n=== Words ===")
print(re.findall(r"\w+", text))   # ['My', 'phone', 'number', ...]

print("\n=== Emails ===")
print(re.findall(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}", text))

# --------------------------------------------------
# === Example 1: Splitting Names ===
# --------------------------------------------------
data1 = {"Full_Name": ["Mr John Doe", "Dr Jane Smith", "Miss Sarah Connor"]}
df1 = pd.DataFrame(data1)
pattern1 = r"(Mr|Ms|Miss|Mrs|Dr|Alhaji|Sis)\s+(\w+)\s+(\w+)"
df1[["Title", "First_Name", "Surname"]] = df1["Full_Name"].str.extract(pattern1)
print("\n=== Names ===")
print(df1)

# --------------------------------------------------
# === Example 2: Phone Numbers ===
# --------------------------------------------------
data2 = {"Phone": ["+123-4567", "+234-9876", "+44-1234"]}
df2 = pd.DataFrame(data2)
pattern2 = r"(\+\d+)\-(\d+)"
df2[["Area_Code", "Line_Number"]] = df2["Phone"].str.extract(pattern2)
print("\n=== Phone Numbers ===")
print(df2)

# --------------------------------------------------
# === Example 3: Addresses ===
# --------------------------------------------------
data3 = {"Address": ["12 King, London", "34 Queen, Paris", "56 Main, Berlin"]}
df3 = pd.DataFrame(data3)
pattern3 = r"(\w+)\s+(\w+)\,\s+(\w+)"
df3[["House_No", "Street", "City"]] = df3["Address"].str.extract(pattern3)
print("\n=== Addresses ===")
print(df3)

# --------------------------------------------------
# === Example 4: Extract Hashtags ===
# --------------------------------------------------
text = "Learning #Python and #Regex is fun! #100DaysOfCode"
print("\n=== Hashtags ===")
print(re.findall(r"#\w+", text))   # ['#Python', '#Regex', '#100DaysOfCode']

# --------------------------------------------------
# === Example 5: Extract Domain Names ===
# --------------------------------------------------
emails = ["user1@gmail.com", "info@csa-research.com", "hello@yahoo.co.uk"]
domains = [re.search(r"@([\w.-]+)", e).group(1) for e in emails]
print("\n=== Domains ===")
print(domains)

# --------------------------------------------------
# === Example 6: Extract Dates ===
# --------------------------------------------------
text = "Events: 12/09/2025, 05/11/2024 and 31/12/2023."
print("\n=== Dates ===")
print(re.findall(r"\d{2}/\d{2}/\d{4}", text))

# --------------------------------------------------
# === Example 7: Words Starting with Capital Letter ===
# --------------------------------------------------
text = "Alice and Bob are Going to Paris in September."
print("\n=== Capital Words ===")
print(re.findall(r"\b[A-Z][a-z]*\b", text))

# --------------------------------------------------
# === Example 8: Extract Quoted Text ===
# --------------------------------------------------
text = 'He said "hello world", then she replied "regex is cool".'
print("\n=== Quoted Text ===")
print(re.findall(r'"(.*?)"', text))

# --------------------------------------------------
# === Example 9: Extract Phone Numbers (Complex) ===
# --------------------------------------------------
text = "Call me at (123) 456-7890 or 987-654-3210."
print("\n=== Complex Phone Numbers ===")
print(re.findall(r"\(?\d{3}\)?[- ]\d{3}[- ]\d{4}", text))

# --------------------------------------------------
# === Example 10: Extract URLs ===
# --------------------------------------------------
text = "Visit https://openai.com or http://csa-research.com for more info."
print("\n=== URLs ===")
print(re.findall(r"https?://[^\s]+", text))

# --------------------------------------------------
# === Example 11: Find Duplicate Words ===
# --------------------------------------------------
text = "This is is a test test sentence."
print("\n=== Duplicate Words ===")
print(re.findall(r"\b(\w+)\s+\1\b", text))

# --------------------------------------------------
# === Example 12: Extract HTML Tags ===
# --------------------------------------------------
html = "<div><p>Hello</p><a href='#'>Link</a></div>"
print("\n=== HTML Tags ===")
print(re.findall(r"<(/?\w+)", html))

# --------------------------------------------------
# === Example 13: Password Validation ===
# --------------------------------------------------
pattern = r"^(?=.*[a-z])(?=.*[A-Z])(?=.*\d)(?=.*[@$!%*?&])[A-Za-z\d@$!%*?&]{8,}$"
passwords = ["Password123!", "weakpass", "StrongPass1@", "NoSpecialChar123"]

print("\n=== Password Validation ===")
for p in passwords:
    if re.match(pattern, p):
        print(f"{p} ✅ Strong")
    else:
        print(f"{p} ❌ Weak")
```
