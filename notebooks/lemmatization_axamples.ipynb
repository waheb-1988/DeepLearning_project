```python
# ==============================================
# NLP Preprocessing: Stopwords, Stemming, Lemmatization
# Using NLTK, WordNet, and spaCy
# ==============================================

import nltk
import spacy
from nltk.corpus import stopwords, wordnet
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag, word_tokenize

# ----------------------------------------------
# Downloads (run once)
# ----------------------------------------------
nltk.download("punkt")
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")
nltk.download("averaged_perceptron_tagger")
nltk.download("averaged_perceptron_tagger_eng")  # Fix for POS tagging

# Load spaCy English model
nlp = spacy.load("en_core_web_sm")

# ----------------------------------------------
# Example Text
# ----------------------------------------------
text = "The children are running quickly towards the playground. They played happily."

# ----------------------------------------------
# 1. Stopword Removal (NLTK)
# ----------------------------------------------
sw = set(stopwords.words("english"))
tokens = word_tokenize(text)
filtered_tokens = [w for w in tokens if w.lower() not in sw]
print("👉 Stopword Removal (NLTK):")
print("Before:", tokens)
print("After :", filtered_tokens)
print("-" * 60)

# ----------------------------------------------
# 2. Stemming (Porter Stemmer)
# ----------------------------------------------
stemmer = PorterStemmer()
stems = [stemmer.stem(w) for w in filtered_tokens]
print("👉 Stemming (PorterStemmer):")
print("Before:", filtered_tokens)
print("After :", stems)
print("-" * 60)

# ----------------------------------------------
# 3. Lemmatization (NLTK + WordNet POS)
# ----------------------------------------------

# Helper to map POS tags
def get_wordnet_pos(tag):
    if tag.startswith("J"):
        return wordnet.ADJ
    elif tag.startswith("V"):
        return wordnet.VERB
    elif tag.startswith("N"):
        return wordnet.NOUN
    elif tag.startswith("R"):
        return wordnet.ADV
    else:
        return wordnet.NOUN

lemmatizer = WordNetLemmatizer()
pos_tags = pos_tag(filtered_tokens)
lemmas_nltk = [lemmatizer.lemmatize(word, get_wordnet_pos(tag)) for word, tag in pos_tags]

print("👉 Lemmatization (NLTK + WordNet):")
print("POS Tags:", pos_tags)
print("After   :", lemmas_nltk)
print("-" * 60)

# ----------------------------------------------
# 4. Lemmatization (spaCy)
# ----------------------------------------------
doc = nlp(text)
lemmas_spacy = [token.lemma_ for token in doc if token.text.lower() not in sw]

print("👉 Lemmatization (spaCy):")
print("Before:", [token.text for token in doc])
print("After :", lemmas_spacy)
print("-" * 60)
```
